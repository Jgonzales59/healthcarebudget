```python

!pip install django.db.backends.mysql
from django.db import models


#We can create our own models here:
class Expense(models.Model):
  CATEGORY_CHOICES = [
    ('Food', 'Food'),
    ('Transportation', 'Transportation'),
    ('Entertainment', 'Entertainment'),
    ('Ultities', 'Ultities'),
    ('Expenses', 'Expenses')
    ('OTHERS', 'OTHERS'),
  ]

  category = models.CharField(max_length=50, choices=CATEGORY_CHOICES)
  amount = models.CharField(max_digits=10, decimal_places=2)
  description = models.TextField(blank=True, null=True)
  date = models.DateTimeField(auto_new_add=True)

  def __str__(self):
    return f"{self.category} - ${self.amount}" #Here Returns our data
```
